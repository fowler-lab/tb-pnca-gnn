# Create `.npy`s for new dataset splits

Load original data from `ds-train.py` and `ds-test.npy`

```python
X={}
Y={}
Z={}

X['train']={}
Y['train']={}
Z['train']={}

with open('../data/inputs/ds-train.npy', 'rb') as f:
    Y['train']['input'] = numpy.load(f)
    X['train']['input'] = numpy.load(f)
    Z['train']['input'] = numpy.load(f, allow_pickle=True)
with open('../data/inputs/ds-test.npy', 'rb') as f:
    Y_test = numpy.load(f)
    X_test = numpy.load(f)
    Z_test = numpy.load(f, allow_pickle=True)

    Y['train']['input'] = numpy.concatenate([Y['train']['input'], Y_test])
    X['train']['input'] = numpy.concatenate([X['train']['input'], X_test])
    Z['train']['input'] = numpy.concatenate([Z['train']['input'], Z_test])
```

```python
# train_df = pandas.read_csv('../data/clustered_2_train_sequences_2.csv')
# test_df = pandas.read_csv('../data/clustered_2_test_sequences_2.csv')
train_df = pandas.read_csv('../data/codon_train_sequences.csv')
test_df = pandas.read_csv('../data/codon_test_sequences.csv')
```

```python
X_new={}
Y_new={}
Z_new={}

X_new['train']={}
Y_new['train']={}
Z_new['train']={}

Y_new['train']['input'] = numpy.array([])
X_new['train']['input'] = numpy.empty((0, X['train']['input'].shape[1]))
Z_new['train']['input'] = numpy.array([])

X_new['test']={}
Y_new['test']={}
Z_new['test']={}

Y_new['test']['input'] = numpy.array([])
X_new['test']['input'] = numpy.empty((0, X['train']['input'].shape[1]))
Z_new['test']['input'] = numpy.array([])
```

```python
for x, y, z, in zip(X['train']['input'], Y['train']['input'], Z['train']['input']):
    # print(x, y, z)
    if z in train_df.MUTATION.values:
        Y_new['train']['input'] = numpy.concatenate([Y_new['train']['input'], [y]])
        X_new['train']['input'] = numpy.concatenate([X_new['train']['input'], [x]])
        Z_new['train']['input'] = numpy.concatenate([Z_new['train']['input'], [z]])
    elif z in test_df.MUTATION.values:
        Y_new['test']['input'] = numpy.concatenate([Y_new['test']['input'], [y]])
        X_new['test']['input'] = numpy.concatenate([X_new['test']['input'], [x]])
        Z_new['test']['input'] = numpy.concatenate([Z_new['test']['input'], [z]])
    else:
        raise Exception
```

```python
# with open('../data/inputs/clustered-ds-train.npy', 'wb') as f:
with open('../data/inputs/codon-ds-train.npy', 'wb') as f:
    numpy.save(f, Y_new['train']['input'])
    numpy.save(f, X_new['train']['input'])
    numpy.save(f, Z_new['train']['input'])

# with open('../data/inputs/clustered-ds-test.npy', 'wb') as f:
with open('../data/inputs/codon-ds-test.npy', 'wb') as f:
    numpy.save(f, Y_new['test']['input'])
    numpy.save(f, X_new['test']['input'])
    numpy.save(f, Z_new['test']['input'])
```
